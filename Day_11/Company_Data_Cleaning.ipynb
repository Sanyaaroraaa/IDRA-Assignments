# Data Cleaning: Company Employee Dataset


```python
import pandas as pd

df = pd.read_csv('Day11_Messy_Company_Employee_Dataset.csv')
df.head()
```

```python
print('Shape:', df.shape)
print('\nData Types:')
print(df.dtypes)

print('\nMissing Values:')
print(df.isnull().sum())

print('\nTotal Missing Values:', df.isnull().sum().sum())
print('\nDuplicate Records:', df.duplicated().sum())
```

```python
for col in ['Department', 'Gender', 'City', 'Work_Mode']:
    print('\n', col)
    print(df[col].value_counts(dropna=False))
```

```python
## Remove Duplicate Records
df = df.drop_duplicates().copy()
print('Rows after removing duplicates:', len(df))
print('Remaining duplicates:', df.duplicated().sum())
```

```python
## Standardize Inconsistent Entries
for col in ['Department', 'Gender', 'City', 'Work_Mode']:
    df[col] = df[col].astype('string').str.strip()

df['Department'] = df['Department'].replace({
    'ENGINEERING': 'Engineering',
    'engineering': 'Engineering'
})

df['Gender'] = df['Gender'].replace({
    'female': 'Female',
    'MALE': 'Male'
})

df['City'] = df['City'].replace({
    'delhi': 'Delhi',
    'DELHI': 'Delhi'
})

df['Work_Mode'] = df['Work_Mode'].replace({
    'REMOTE': 'Remote',
    'remote': 'Remote'
})
```

```python
## Correct Data Types
numeric_cols = ['Age', 'Annual_Salary', 'Experience_Years', 'Performance_Score']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['Joining_Date'] = pd.to_datetime(df['Joining_Date'], errors='coerce')

print(df.dtypes)
```

```python
## Handle Missing Values
numeric_impute_cols = ['Age', 'Annual_Salary', 'Experience_Years', 'Performance_Score']

for col in numeric_impute_cols:
    df[col] = df[col].fillna(df[col].median())

categorical_impute_cols = ['Department', 'Gender', 'City', 'Work_Mode']

for col in categorical_impute_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())
```

```python
df['Joining_Date'] = df['Joining_Date'].dt.strftime('%Y-%m-%d')
df['Age'] = df['Age'].round().astype(int)
df['Annual_Salary'] = df['Annual_Salary'].round(2)
df['Experience_Years'] = df['Experience_Years'].round(1)
df['Performance_Score'] = df['Performance_Score'].round().astype(int)
```

```python
## Verify After Cleaning
print('Final Shape:', df.shape)
print('\nMissing Values:')
print(df.isnull().sum())
print('\nTotal Missing Values:', df.isnull().sum().sum())
print('Duplicate Records:', df.duplicated().sum())
print('\nFinal Data Types:')
print(df.dtypes)
df.head()
```
