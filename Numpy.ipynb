```python
import numpy as np
```

```python
## 1. Creating a 1D Array
arr1 = np.array([10, 20, 30, 40, 50])

print(arr1)
```

```python
## 2. Creating a 2D Array
arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr2)
```

```python
## 3. Creating a 3D Array
arr3 = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print(arr3)
```

```python
## 4. Checking Dimensions and Data Types
print("Dimension of arr1:", arr1.ndim)
print("Dimension of arr2:", arr2.ndim)
print("Dimension of arr3:", arr3.ndim)

print("\nShape of arr1:", arr1.shape)
print("Shape of arr2:", arr2.shape)
print("Shape of arr3:", arr3.shape)

print("\nData type of arr1:", arr1.dtype)
```

```python
## 5. Indexing
numbers = np.array([10, 20, 30, 40, 50])

print("First element:", numbers[0])
print("Third element:", numbers[2])
print("Last element:", numbers[-1])
```

```python
## 6. 2D Indexing
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Element at row 1, column 2:", matrix[0, 1])
print("Element at row 2, column 3:", matrix[1, 2])
```

```python
## 7. Slicing
numbers = np.array([10, 20, 30, 40, 50, 60])

print("First three elements:", numbers[:3])
print("Elements from index 2:", numbers[2:])
print("Every second element:", numbers[::2])
```

```python
## 8. 2D Slicing
matrix = np.array([
    [1, 2, 3],
    [9, 5, 8],
    [7, 8, 4]
])

print("First two rows:")
print(matrix[:2])

print("\nFirst two columns:")
print(matrix[:, :2])
```

```python
## 9. Reshaping
arr = np.arange(1, 13)

print("Original array:")
print(arr)

new_arr = arr.reshape(3, 4)

print("\nAfter reshaping:")
print(new_arr)
```

```python
## 10. Creating Arrays Using NumPy Functions
# zeros()
zero_array = np.zeros((2, 3))
print("Zeros:")
print(zero_array)

# ones()
one_array = np.ones((2, 3))
print("\nOnes:")
print(one_array)

# arange()
range_array = np.arange(1, 11)
print("\nArange:")
print(range_array)

# linspace()
line_array = np.linspace(0, 10, 5)
print("\nLinspace:")
print(line_array)
```

```python
## 11. Mathematical Operations
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
## 12. Vectorized Operations
numbers = np.array([1, 2, 3, 4, 5])

print("Original:", numbers)
print("Adding 10:", numbers + 10)
print("Multiplying by 2:", numbers * 2)
print("Square:", numbers ** 2)
```

```python
## 13. Boolean Masking
marks = np.array([45, 67, 32, 89, 76, 55, 91])

print("Marks greater than 70:")
print(marks[marks > 70])
```

```python
print("Marks below 50:")
print(marks[marks < 50])
```

```python
## 14. Broadcasting
prices = np.array([100, 200, 300, 400])

new_prices = prices + 10

print("Original prices:")
print(prices)

print("\nAfter adding 10:")
print(new_prices)
```

```python
## 15. Broadcasting with 2D Array
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

values = np.array([10, 20, 30])

result = matrix + values

print(result)
```

```python
## 16. Aggregate Functions
numbers = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Minimum:", np.min(numbers))
print("Maximum:", np.max(numbers))
print("Median:", np.median(numbers))
```

```python
## 17. Small Practice Example - Student Marks
marks = np.array([78, 85, 62, 90, 74, 88, 95, 69])

print("Student marks:")
print(marks)

print("\nAverage marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
print("Median marks:", np.median(marks))

print("\nStudents scoring above 75:")
print(marks[marks > 75])
```
