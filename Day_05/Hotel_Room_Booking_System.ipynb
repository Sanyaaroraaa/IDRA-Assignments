```python
import csv
from datetime import datetime


# ---------------- ROOM CLASS ----------------

class Room:

    def __init__(self, room_no, room_type, price, facilities, floor):
        self.room_no = room_no
        self.room_type = room_type
        self.price = price
        self.facilities = facilities
        self.floor = floor
        self.status = "Available"

    def show_room(self):

        print("\nRoom:", self.room_no)
        print("Type:", self.room_type)
        print("Price: Rs.", self.price, "per night")
        print("Floor:", self.floor)
        print("Status:", self.status)
        print("Facilities:", ", ".join(self.facilities))


# ---------------- CUSTOMER CLASS ----------------

class Customer:

    def __init__(self, customer_id, name, phone, email, student):
        self.customer_id = customer_id
        self.name = name
        self.phone = phone
        self.email = email
        self.student = student


# ---------------- BOOKING CLASS ----------------

class Booking:

    def __init__(
        self,
        booking_id,
        customer_id,
        room_no,
        check_in,
        check_out,
        nights,
        guests,
        package,
        services,
        request,
        total,
        status
    ):

        self.booking_id = booking_id
        self.customer_id = customer_id
        self.room_no = room_no
        self.check_in = check_in
        self.check_out = check_out
        self.nights = nights
        self.guests = guests
        self.package = package
        self.services = services
        self.request = request
        self.total = total
        self.status = status


# ---------------- HOTEL CLASS ----------------

class Hotel:

    def __init__(self):

        self.rooms = []
        self.customers = []
        self.bookings = []

        # Dictionary
        self.services = {
            "Breakfast": 200,
            "Room Service": 150,
            "Laundry": 100,
            "Extra Bed": 500,
            "Airport Pickup": 800
        }

        # Dictionary
        self.packages = {
            "Basic": 0,
            "Family": 500,
            "Couple": 700,
            "Business": 600
        }

        self.load_rooms()
        self.load_customers()
        self.load_bookings()

        if len(self.rooms) == 0:
            self.add_default_rooms()

        self.update_room_status()


    # ---------------- ROOM FUNCTIONS ----------------

    def add_default_rooms(self):

        self.rooms = [

            Room(
                101,
                "Single",
                1500,
                ("Wi-Fi", "TV"),
                1
            ),

            Room(
                102,
                "Single",
                1500,
                ("Wi-Fi", "TV"),
                1
            ),

            Room(
                201,
                "Double",
                2500,
                ("Wi-Fi", "TV", "Mini Fridge"),
                2
            ),

            Room(
                202,
                "Double",
                2500,
                ("Wi-Fi", "TV", "Mini Fridge"),
                2
            ),

            Room(
                301,
                "Deluxe",
                3500,
                ("Wi-Fi", "TV", "AC", "Mini Fridge"),
                3
            ),

            Room(
                302,
                "Deluxe",
                3500,
                ("Wi-Fi", "TV", "AC", "Mini Fridge"),
                3
            ),

            Room(
                401,
                "Suite",
                5000,
                (
                    "Wi-Fi",
                    "TV",
                    "AC",
                    "Mini Fridge",
                    "Living Room"
                ),
                4
            ),

            Room(
                402,
                "Suite",
                5000,
                (
                    "Wi-Fi",
                    "TV",
                    "AC",
                    "Mini Fridge",
                    "Living Room"
                ),
                4
            )
        ]

        self.save_rooms()


    def find_room(self, room_no):

        for room in self.rooms:

            if room.room_no == room_no:
                return room

        return None


    def view_rooms(self):

        print("\n--------- HOTEL ROOMS ---------")

        if len(self.rooms) == 0:

            print("No rooms available.")
            return

        for room in self.rooms:

            room.show_room()
            print("-----------------------------")


    def update_room_status(self):

        # First make every room available
        for room in self.rooms:
            room.status = "Available"

        # Set is used to keep track of occupied rooms
        occupied_rooms = set()

        for booking in self.bookings:

            if booking.status in ["Confirmed", "Checked In"]:

                occupied_rooms.add(booking.room_no)

        for room in self.rooms:

            if room.room_no in occupied_rooms:
                room.status = "Booked"


    # ---------------- ROOM RECOMMENDATION ----------------

    def recommend_room(self):

        print("\n================================")
        print("       ROOM RECOMMENDATION")
        print("================================")

        print("Tell us what you are looking for.")

        print("\nWhat is your budget per night?")
        print("1. Below Rs. 2000")
        print("2. Rs. 2000 - Rs. 3000")
        print("3. Rs. 3000 - Rs. 4000")
        print("4. Above Rs. 4000")

        budget = input("Enter choice: ")

        if budget not in ["1", "2", "3", "4"]:

            print("Invalid budget choice.")
            return

        print("\nWhat kind of stay do you want?")
        print("1. Simple")
        print("2. Family")
        print("3. Comfortable")
        print("4. Luxury")

        stay = input("Enter choice: ")

        if stay not in ["1", "2", "3", "4"]:

            print("Invalid stay choice.")
            return

        try:

            guests = int(
                input("\nHow many guests? ")
            )

            if guests <= 0:

                print("Number of guests must be greater than 0.")
                return

        except ValueError:

            print("Please enter a valid number.")
            return

        best_room = None
        best_score = -1
        best_reasons = []

        for room in self.rooms:

            if room.status != "Available":
                continue

            score = 0
            reasons = []

            # -------- BUDGET SCORE --------

            if budget == "1" and room.price < 2000:

                score += 2
                reasons.append("Fits your budget")

            elif budget == "2" and 2000 <= room.price <= 3000:

                score += 2
                reasons.append("Fits your budget")

            elif budget == "3" and 3000 < room.price <= 4000:

                score += 2
                reasons.append("Fits your budget")

            elif budget == "4" and room.price > 4000:

                score += 2
                reasons.append("Fits your budget")


            # -------- STAY TYPE SCORE --------

            if stay == "1" and room.room_type == "Single":

                score += 2
                reasons.append("Good for a simple stay")

            elif stay == "2" and room.room_type == "Double":

                score += 2
                reasons.append("Good for a family stay")

            elif stay == "3" and room.room_type == "Deluxe":

                score += 2
                reasons.append("Good for a comfortable stay")

            elif stay == "4" and room.room_type == "Suite":

                score += 2
                reasons.append("Good for a luxury stay")


            # -------- GUEST SCORE --------

            capacity = 0

            if room.room_type == "Single":
                capacity = 2

            elif room.room_type == "Double":
                capacity = 4

            elif room.room_type == "Deluxe":
                capacity = 4

            elif room.room_type == "Suite":
                capacity = 6


            if guests <= capacity:

                score += 2
                reasons.append("Suitable for your group")

            else:

                continue


            # -------- FACILITY BONUS --------

            if len(room.facilities) >= 4:

                score += 1
                reasons.append("Has extra facilities")


            # Save the best room

            if score > best_score:

                best_score = score
                best_room = room
                best_reasons = reasons


        if best_room is None:

            print("\nSorry, no room can accommodate your group.")

            return


        print("\n--------------------------------")
        print("        OUR RECOMMENDATION")
        print("--------------------------------")

        print("Room:", best_room.room_no)
        print("Type:", best_room.room_type)
        print("Price: Rs.", best_room.price, "per night")
        print("Floor:", best_room.floor)

        print(
            "Facilities:",
            ", ".join(best_room.facilities)
        )

        print("\nWhy we recommend it:")

        for reason in best_reasons:

            print("✓", reason)

        print("\nMatch Score:", best_score, "/ 7")

        if best_score >= 6:

            print("This room is a great match for you!")

        elif best_score >= 4:

            print("This room matches most of your preferences.")

        else:

            print(
                "This is the closest available option "
                "based on your choices."
            )


    # ---------------- SEARCH ROOMS ----------------

    def search_rooms(self):

        print("\n--------- SEARCH ROOMS ---------")

        print("1. Search by room type")
        print("2. Search by maximum price")

        choice = input("Enter choice: ")

        if choice == "1":

            room_type = input(
                "Enter room type: "
            ).strip().lower()

            found = False

            for room in self.rooms:

                if (
                    room.room_type.lower() == room_type
                    and room.status == "Available"
                ):

                    print(
                        "Room",
                        room.room_no,
                        "-",
                        room.room_type,
                        "- Rs.",
                        room.price
                    )

                    found = True

            if not found:

                print("No matching available room found.")


        elif choice == "2":

            try:

                max_price = float(
                    input("Enter maximum price: ")
                )

                if max_price <= 0:

                    print("Price must be greater than 0.")
                    return

                found = False

                for room in self.rooms:

                    if (
                        room.price <= max_price
                        and room.status == "Available"
                    ):

                        print(
                            "Room",
                            room.room_no,
                            "-",
                            room.room_type,
                            "- Rs.",
                            room.price
                        )

                        found = True

                if not found:

                    print("No room found within this budget.")

            except ValueError:

                print("Please enter a valid amount.")


        else:

            print("Invalid choice.")


    # ---------------- CUSTOMER FUNCTIONS ----------------

    def find_customer(self, customer_id):

        for customer in self.customers:

            if customer.customer_id == customer_id:

                return customer

        return None


    def register_customer(self):

        print("\n--------- CUSTOMER REGISTRATION ---------")

        name = input("Enter name: ").strip()

        if name == "":

            print("Name cannot be empty.")
            return


        phone = input("Enter phone number: ").strip()

        if not phone.isdigit() or len(phone) != 10:

            print("Enter a valid 10 digit phone number.")
            return


        # Prevent duplicate phone numbers

        for customer in self.customers:

            if customer.phone == phone:

                print("This phone number is already registered.")

                print(
                    "Your Customer ID is:",
                    customer.customer_id
                )

                return


        email = input("Enter email: ").strip()

        if "@" not in email or "." not in email:

            print("Please enter a valid email.")
            return


        student = input(
            "Are you a student? (y/n): "
        ).lower()

        if student == "y":

            student = "Yes"

        elif student == "n":

            student = "No"

        else:

            print("Please enter y or n.")
            return


        customer_id = self.generate_customer_id()

        customer = Customer(
            customer_id,
            name,
            phone,
            email,
            student
        )

        self.customers.append(customer)

        self.save_customers()

        print("\nCustomer registered successfully.")
        print("Your Customer ID is:", customer_id)


    def generate_customer_id(self):

        highest = 0

        for customer in self.customers:

            try:

                number = int(
                    customer.customer_id[1:]
                )

                if number > highest:
                    highest = number

            except ValueError:

                pass

        return "C" + str(highest + 1).zfill(3)


    # ---------------- SERVICES ----------------

    def show_services(self):

        print("\n--------- EXTRA SERVICES ---------")

        service_names = list(
            self.services.keys()
        )

        for i in range(len(service_names)):

            service = service_names[i]

            print(
                i + 1,
                ".",
                service,
                "- Rs.",
                self.services[service]
            )

        print("0. Done")


    def choose_services(self):

        selected = {}

        while True:

            self.show_services()

            try:

                choice = int(
                    input("Choose service: ")
                )

            except ValueError:

                print("Please enter a number.")
                continue


            if choice == 0:

                break


            service_names = list(
                self.services.keys()
            )

            if choice < 1 or choice > len(service_names):

                print("Invalid service choice.")
                continue


            service = service_names[choice - 1]


            try:

                quantity = int(
                    input("Enter quantity: ")
                )

                if quantity <= 0:

                    print("Quantity must be greater than 0.")
                    continue

            except ValueError:

                print("Enter a valid quantity.")
                continue


            if service in selected:

                selected[service] += quantity

            else:

                selected[service] = quantity


            print(service, "added.")


        return selected


    # ---------------- PACKAGES ----------------

    def choose_package(self):

        print("\n--------- STAY PACKAGES ---------")

        print("1. Basic - Rs. 0")
        print("2. Family - Rs. 500")
        print("3. Couple - Rs. 700")
        print("4. Business - Rs. 600")

        choice = input("Choose package: ")

        if choice == "1":

            return "Basic"

        elif choice == "2":

            return "Family"

        elif choice == "3":

            return "Couple"

        elif choice == "4":

            return "Business"

        else:

            print("Invalid choice. Basic package selected.")

            return "Basic"


    # ---------------- BOOK ROOM ----------------

    def book_room(self):

        print("\n================================")
        print("           BOOK ROOM")
        print("================================")


        if len(self.rooms) == 0:

            print("No rooms are available.")
            return


        customer_id = input(
            "Enter Customer ID: "
        ).upper()

        customer = self.find_customer(
            customer_id
        )

        if customer is None:

            print("Customer not found.")
            print("Please register first.")

            return


        self.update_room_status()

        self.view_rooms()


        try:

            room_no = int(
                input("\nEnter room number: ")
            )

        except ValueError:

            print("Enter a valid room number.")
            return


        room = self.find_room(room_no)


        if room is None:

            print("Room does not exist.")
            return


        if room.status != "Available":

            print("This room is not available.")
            return


        # -------- DATES --------

        check_in = input(
            "Enter check-in date (DD-MM-YYYY): "
        ).strip()

        check_out = input(
            "Enter check-out date (DD-MM-YYYY): "
        ).strip()


        try:

            date1 = datetime.strptime(
                check_in,
                "%d-%m-%Y"
            )

            date2 = datetime.strptime(
                check_out,
                "%d-%m-%Y"
            )

            today = datetime.now().date()

            if date1.date() < today:

                print("Check-in date cannot be in the past.")
                return


            nights = (date2 - date1).days


            if nights <= 0:

                print(
                    "Check-out must be after check-in."
                )

                return


        except ValueError:

            print(
                "Invalid date. Please use DD-MM-YYYY."
            )

            return


        # -------- GUESTS --------

        try:

            guests = int(
                input("Number of guests: ")
            )

            if guests <= 0:

                print("Invalid number of guests.")
                return

        except ValueError:

            print("Enter a valid number.")
            return


        capacity = {

            "Single": 2,
            "Double": 4,
            "Deluxe": 4,
            "Suite": 6

        }


        if guests > capacity[room.room_type]:

            print(
                "This room can accommodate only",
                capacity[room.room_type],
                "guests."
            )

            return


        # -------- PACKAGE --------

        package = self.choose_package()


        # -------- SERVICES --------

        services = self.choose_services()


        # -------- SPECIAL REQUEST --------

        request = input(
            "\nAny special request? "
            "(Press Enter if none): "
        ).strip()


        # -------- BILL CALCULATION --------

        room_amount = room.price * nights

        package_amount = self.packages[package]

        service_amount = 0


        for service in services:

            service_amount += (
                self.services[service]
                * services[service]
            )


        subtotal = (
            room_amount
            + package_amount
            + service_amount
        )


        discount = 0


        # Student discount

        if customer.student == "Yes":

            discount += subtotal * 0.05


        # Coupon

        coupon = input(
            "\nEnter coupon code "
            "(Press Enter to skip): "
        ).strip().upper()


        if coupon == "HOTEL10":

            discount += subtotal * 0.10

            print("10% coupon applied.")

        elif coupon == "STAY5":

            discount += subtotal * 0.05

            print("5% coupon applied.")

        elif coupon == "WELCOME":

            discount += 200

            print("Rs. 200 coupon applied.")

        elif coupon != "":

            print("Invalid coupon.")


        if discount > subtotal:

            discount = subtotal


        after_discount = subtotal - discount

        gst = after_discount * 0.18

        total = after_discount + gst


        # -------- BOOKING ID --------

        booking_id = self.generate_booking_id()


        booking = Booking(

            booking_id,
            customer_id,
            room_no,
            check_in,
            check_out,
            nights,
            guests,
            package,
            services,
            request,
            round(total, 2),
            "Confirmed"

        )


        self.bookings.append(booking)

        room.status = "Booked"


        self.save_bookings()
        self.save_rooms()


        self.print_bill(

            booking,
            customer,
            room,
            room_amount,
            package_amount,
            service_amount,
            discount,
            gst

        )


    def generate_booking_id(self):

        highest = 0

        for booking in self.bookings:

            try:

                number = int(
                    booking.booking_id[1:]
                )

                if number > highest:
                    highest = number

            except ValueError:

                pass

        return "B" + str(highest + 1).zfill(4)


    # ---------------- BILL ----------------

    def print_bill(
        self,
        booking,
        customer,
        room,
        room_amount,
        package_amount,
        service_amount,
        discount,
        gst
    ):

        print("\n")
        print("----------------------------------------")
        print("             HOTEL SUNRISE")
        print("              BOOKING BILL")
        print("----------------------------------------")

        print("Booking ID:", booking.booking_id)
        print("Customer:", customer.name)
        print("Room:", room.room_no)
        print("Room Type:", room.room_type)
        print("Check-in:", booking.check_in)
        print("Check-out:", booking.check_out)
        print("Nights:", booking.nights)
        print("Guests:", booking.guests)
        print("Package:", booking.package)

        if booking.services:

            print("\nExtra Services:")

            for service in booking.services:

                print(
                    service,
                    "x",
                    booking.services[service]
                )

        if booking.request != "":

            print(
                "\nSpecial Request:",
                booking.request
            )

        print("----------------------------------------")

        print(
            "Room charges: Rs.",
            round(room_amount, 2)
        )

        print(
            "Package: Rs.",
            round(package_amount, 2)
        )

        print(
            "Services: Rs.",
            round(service_amount, 2)
        )

        print(
            "Discount: Rs.",
            round(discount, 2)
        )

        print(
            "GST: Rs.",
            round(gst, 2)
        )

        print("----------------------------------------")

        print(
            "TOTAL: Rs.",
            booking.total
        )

        print(
            "Status:",
            booking.status
        )

        print("----------------------------------------")


    # ---------------- BOOKING SEARCH ----------------

    def find_booking(self, booking_id):

        for booking in self.bookings:

            if booking.booking_id == booking_id:

                return booking

        return None


    # ---------------- VIEW BOOKING ----------------

    def view_booking(self):

        booking_id = input(
            "Enter Booking ID: "
        ).upper()

        booking = self.find_booking(
            booking_id
        )


        if booking is None:

            print("Booking not found.")
            return


        customer = self.find_customer(
            booking.customer_id
        )

        room = self.find_room(
            booking.room_no
        )


        print("\n--------- BOOKING DETAILS ---------")


        print(
            "Booking ID:",
            booking.booking_id
        )


        if customer:

            print(
                "Customer:",
                customer.name
            )


        print(
            "Room:",
            booking.room_no
        )


        if room:

            print(
                "Room Type:",
                room.room_type
            )


        print(
            "Check-in:",
            booking.check_in
        )

        print(
            "Check-out:",
            booking.check_out
        )

        print(
            "Nights:",
            booking.nights
        )

        print(
            "Guests:",
            booking.guests
        )

        print(
            "Package:",
            booking.package
        )

        print(
            "Services:",
            booking.services
        )

        print(
            "Special Request:",
            booking.request
        )

        print(
            "Total: Rs.",
            booking.total
        )

        print(
            "Status:",
            booking.status
        )


    # ---------------- BOOKING HISTORY ----------------

    def booking_history(self):

        customer_id = input(
            "Enter Customer ID: "
        ).upper()

        customer = self.find_customer(
            customer_id
        )


        if customer is None:

            print("Customer not found.")
            return


        found = False


        print(
            "\n--------- BOOKING HISTORY ---------"
        )


        for booking in self.bookings:

            if booking.customer_id == customer_id:

                print(
                    "\nBooking:",
                    booking.booking_id
                )

                print(
                    "Room:",
                    booking.room_no
                )

                print(
                    "Dates:",
                    booking.check_in,
                    "to",
                    booking.check_out
                )

                print(
                    "Total: Rs.",
                    booking.total
                )

                print(
                    "Status:",
                    booking.status
                )

                found = True


        if not found:

            print("No bookings found for this customer.")


    # ---------------- CANCEL BOOKING ----------------

    def cancel_booking(self):

        booking_id = input(
            "Enter Booking ID: "
        ).upper()

        booking = self.find_booking(
            booking_id
        )


        if booking is None:

            print("Booking not found.")
            return


        if booking.status == "Cancelled":

            print("Booking is already cancelled.")
            return


        if booking.status == "Checked Out":

            print("This booking has already been completed.")
            return


        confirm = input(
            "Cancel this booking? (y/n): "
        ).lower()


        if confirm != "y":

            print("Booking was not cancelled.")
            return


        cancellation_charge = (
            booking.total * 0.10
        )

        refund = (
            booking.total
            - cancellation_charge
        )


        booking.status = "Cancelled"


        room = self.find_room(
            booking.room_no
        )


        if room:

            room.status = "Available"


        self.save_bookings()
        self.save_rooms()


        print("\nBooking cancelled successfully.")

        print(
            "Cancellation charge: Rs.",
            round(cancellation_charge, 2)
        )

        print(
            "Refund amount: Rs.",
            round(refund, 2)
        )


    # ---------------- CHECK IN ----------------

    def check_in(self):

        booking_id = input(
            "Enter Booking ID: "
        ).upper()

        booking = self.find_booking(
            booking_id
        )


        if booking is None:

            print("Booking not found.")
            return


        if booking.status != "Confirmed":

            print(
                "Only confirmed bookings can be checked in."
            )

            return


        booking.status = "Checked In"

        self.save_bookings()

        print(
            "Customer checked in successfully."
        )


    # ---------------- CHECK OUT ----------------

    def check_out(self):

        booking_id = input(
            "Enter Booking ID: "
        ).upper()

        booking = self.find_booking(
            booking_id
        )


        if booking is None:

            print("Booking not found.")
            return


        if booking.status != "Checked In":

            print(
                "Customer must be checked in before checkout."
            )

            return


        booking.status = "Checked Out"


        room = self.find_room(
            booking.room_no
        )


        if room:

            room.status = "Available"


        self.save_bookings()
        self.save_rooms()


        print("\nCheckout completed successfully.")
        print("Thank you for staying with us!")


    # ---------------- OFFERS ----------------

    def show_offers(self):

        print("\n--------- CURRENT OFFERS ---------")

        print("HOTEL10  - 10% discount")
        print("STAY5    - 5% discount")
        print("WELCOME  - Rs. 200 discount")
        print("Students - Extra 5% discount")


    # ---------------- ADMIN FUNCTIONS ----------------

    def view_customers(self):

        print("\n--------- CUSTOMERS ---------")


        if len(self.customers) == 0:

            print("No customers registered.")
            return


        for customer in self.customers:

            print(
                customer.customer_id,
                "-",
                customer.name,
                "-",
                customer.phone,
                "-",
                customer.email
            )


    def view_all_bookings(self):

        print("\n--------- ALL BOOKINGS ---------")


        if len(self.bookings) == 0:

            print("No bookings yet.")
            return


        for booking in self.bookings:

            print(
                booking.booking_id,
                "| Customer:",
                booking.customer_id,
                "| Room:",
                booking.room_no,
                "| Rs.",
                booking.total,
                "|",
                booking.status
            )


    # ---------------- HOTEL STATISTICS ----------------

    def hotel_statistics(self):

        self.update_room_status()


        # Set
        occupied_rooms = set()


        for room in self.rooms:

            if room.status != "Available":

                occupied_rooms.add(
                    room.room_no
                )


        total_revenue = 0

        room_types = {}


        for booking in self.bookings:

            if booking.status != "Cancelled":

                total_revenue += booking.total


                room = self.find_room(
                    booking.room_no
                )


                if room:

                    if room.room_type in room_types:

                        room_types[
                            room.room_type
                        ] += 1

                    else:

                        room_types[
                            room.room_type
                        ] = 1


        print(
            "\n--------- HOTEL STATISTICS ---------"
        )


        print(
            "Total rooms:",
            len(self.rooms)
        )

        print(
            "Available rooms:",
            len(self.rooms)
            - len(occupied_rooms)
        )

        print(
            "Occupied rooms:",
            len(occupied_rooms)
        )

        print(
            "Total customers:",
            len(self.customers)
        )

        print(
            "Total bookings:",
            len(self.bookings)
        )

        print(
            "Total revenue: Rs.",
            round(total_revenue, 2)
        )


        if len(room_types) > 0:

            most_booked = ""
            highest = 0


            for room_type in room_types:

                if room_types[room_type] > highest:

                    highest = room_types[room_type]

                    most_booked = room_type


            print(
                "Most booked room:",
                most_booked
            )


    # ---------------- ADD ROOM ----------------

    def add_room(self):

        print("\n--------- ADD NEW ROOM ---------")


        try:

            room_no = int(
                input("Enter room number: ")
            )

        except ValueError:

            print("Enter a valid room number.")
            return


        if self.find_room(room_no):

            print("Room already exists.")
            return


        print("\n1. Single")
        print("2. Double")
        print("3. Deluxe")
        print("4. Suite")


        choice = input("Choose room type: ")


        if choice == "1":

            room_type = "Single"

            price = 1500

            facilities = (
                "Wi-Fi",
                "TV"
            )


        elif choice == "2":

            room_type = "Double"

            price = 2500

            facilities = (
                "Wi-Fi",
                "TV",
                "Mini Fridge"
            )


        elif choice == "3":

            room_type = "Deluxe"

            price = 3500

            facilities = (
                "Wi-Fi",
                "TV",
                "AC",
                "Mini Fridge"
            )


        elif choice == "4":

            room_type = "Suite"

            price = 5000

            facilities = (
                "Wi-Fi",
                "TV",
                "AC",
                "Mini Fridge",
                "Living Room"
            )


        else:

            print("Invalid room type.")
            return


        floor = room_no // 100


        room = Room(
            room_no,
            room_type,
            price,
            facilities,
            floor
        )


        self.rooms.append(room)

        self.save_rooms()


        print("Room added successfully.")


    # ---------------- FILE HANDLING ----------------

    def save_rooms(self):

        with open(
            "rooms.csv",
            "w",
            newline=""
        ) as file:

            writer = csv.writer(file)


            writer.writerow([
                "room_no",
                "room_type",
                "price",
                "facilities",
                "floor",
                "status"
            ])


            for room in self.rooms:

                writer.writerow([
                    room.room_no,
                    room.room_type,
                    room.price,
                    "|".join(room.facilities),
                    room.floor,
                    room.status
                ])


    def load_rooms(self):

        try:

            with open(
                "rooms.csv",
                "r",
                newline=""
            ) as file:

                reader = csv.DictReader(file)


                for row in reader:

                    facilities = tuple(
                        row["facilities"].split("|")
                    )


                    room = Room(
                        int(row["room_no"]),
                        row["room_type"],
                        float(row["price"]),
                        facilities,
                        int(row["floor"])
                    )


                    room.status = row["status"]

                    self.rooms.append(room)


        except FileNotFoundError:

            pass


    def save_customers(self):

        with open(
            "customers.csv",
            "w",
            newline=""
        ) as file:

            writer = csv.writer(file)


            writer.writerow([
                "customer_id",
                "name",
                "phone",
                "email",
                "student"
            ])


            for customer in self.customers:

                writer.writerow([
                    customer.customer_id,
                    customer.name,
                    customer.phone,
                    customer.email,
                    customer.student
                ])


    def load_customers(self):

        try:

            with open(
                "customers.csv",
                "r",
                newline=""
            ) as file:

                reader = csv.DictReader(file)


                for row in reader:

                    customer = Customer(
                        row["customer_id"],
                        row["name"],
                        row["phone"],
                        row["email"],
                        row["student"]
                    )


                    self.customers.append(customer)


        except FileNotFoundError:

            pass


    def save_bookings(self):

        with open(
            "bookings.csv",
            "w",
            newline=""
        ) as file:

            writer = csv.writer(file)


            writer.writerow([
                "booking_id",
                "customer_id",
                "room_no",
                "check_in",
                "check_out",
                "nights",
                "guests",
                "package",
                "services",
                "request",
                "total",
                "status"
            ])


            for booking in self.bookings:

                service_text = ""


                for service in booking.services:

                    service_text += (
                        service
                        + ":"
                        + str(
                            booking.services[service]
                        )
                        + "|"
                    )


                writer.writerow([
                    booking.booking_id,
                    booking.customer_id,
                    booking.room_no,
                    booking.check_in,
                    booking.check_out,
                    booking.nights,
                    booking.guests,
                    booking.package,
                    service_text,
                    booking.request,
                    booking.total,
                    booking.status
                ])


    def load_bookings(self):

        try:

            with open(
                "bookings.csv",
                "r",
                newline=""
            ) as file:

                reader = csv.DictReader(file)


                for row in reader:

                    services = {}


                    if row["services"] != "":

                        parts = row["services"].split("|")


                        for part in parts:

                            if ":" in part:

                                name, quantity = part.split(":")

                                services[name] = int(
                                    quantity
                                )


                    booking = Booking(

                        row["booking_id"],

                        row["customer_id"],

                        int(row["room_no"]),

                        row["check_in"],

                        row["check_out"],

                        int(row["nights"]),

                        int(row["guests"]),

                        row["package"],

                        services,

                        row["request"],

                        float(row["total"]),

                        row["status"]

                    )


                    self.bookings.append(booking)


        except FileNotFoundError:

            pass


# ---------------- ADMIN MENU ----------------

def admin_menu(hotel):

    password = input(
        "\nEnter admin password: "
    )


    if password != "admin123":

        print("Wrong password.")
        return


    while True:

        print("\n--------- ADMIN PANEL ---------")

        print("1. View all rooms")
        print("2. View all customers")
        print("3. View all bookings")
        print("4. Add room")
        print("5. Check-in customer")
        print("6. Check-out customer")
        print("7. Hotel statistics")
        print("8. Back")


        choice = input("Enter choice: ")


        if choice == "1":

            hotel.view_rooms()


        elif choice == "2":

            hotel.view_customers()


        elif choice == "3":

            hotel.view_all_bookings()


        elif choice == "4":

            hotel.add_room()


        elif choice == "5":

            hotel.check_in()


        elif choice == "6":

            hotel.check_out()


        elif choice == "7":

            hotel.hotel_statistics()


        elif choice == "8":

            break


        else:

            print("Invalid choice.")


# ---------------- MAIN MENU ----------------

def main():

    hotel = Hotel()


    while True:

        print("\n======================================")
        print("       HOTEL ROOM BOOKING SYSTEM")
        print("======================================")

        print("1. Register Customer")
        print("2. View Rooms")
        print("3. Search Rooms")
        print("4. Find My Recommended Room")
        print("5. Book a Room")
        print("6. View Booking")
        print("7. Booking History")
        print("8. Cancel Booking")
        print("9. View Offers")
        print("10. Admin Panel")
        print("11. Exit")


        choice = input("\nEnter choice: ")


        if choice == "1":

            hotel.register_customer()


        elif choice == "2":

            hotel.view_rooms()


        elif choice == "3":

            hotel.search_rooms()


        elif choice == "4":

            hotel.recommend_room()


        elif choice == "5":

            hotel.book_room()


        elif choice == "6":

            hotel.view_booking()


        elif choice == "7":

            hotel.booking_history()


        elif choice == "8":

            hotel.cancel_booking()


        elif choice == "9":

            hotel.show_offers()


        elif choice == "10":

            admin_menu(hotel)


        elif choice == "11":

            print(
                "\nThank you for using "
                "Hotel Room Booking System!"
            )

            break


        else:

            print(
                "Invalid choice. Please select "
                "a number from 1 to 11."
            )


main()\
```
