## Clean Delay Data for Delay-Specific Analysis




```python
delay_df = df.dropna(subset=["Delay_Minutes"]).copy()

print("Original rows:", len(df))
print("Rows with recorded delay:", len(delay_df))
print("Rows removed for delay analysis:", len(df) - len(delay_df))
```

## Transform — Create Useful Columns



```python
# Extract month from the datetime column
df["Flight_Month"] = df["Flight_Date"].dt.month

# Occupancy rate
df["Occupancy_Rate"] = (
    df["Passengers"] / df["Seat_Capacity"] * 100
).round(2)

df[[
    "Flight_ID",
    "Flight_Date",
    "Flight_Month",
    "Passengers",
    "Seat_Capacity",
    "Occupancy_Rate"
]].head(10)
```

### Create a delay category using `apply()`




```python
def delay_category(minutes):
    if pd.isna(minutes):
        return "Not Recorded"
    elif minutes == 0:
        return "On Time"
    elif minutes <= 30:
        return "Short Delay"
    elif minutes <= 60:
        return "Long Delay"
    else:
        return "Very Long Delay"

df["Delay_Category"] = df["Delay_Minutes"].apply(delay_category)

df[[
    "Flight_ID",
    "Delay_Minutes",
    "Delay_Category"
]].head(15)
```

### Use `map()` for a simple transformation




```python
status_priority = {
    "On Time": 1,
    "Delayed": 2,
    "Cancelled": 3
}

df["Status_Priority"] = df["Flight_Status"].map(status_priority)

df[[
    "Flight_ID",
    "Flight_Status",
    "Status_Priority"
]].head(10)
```

##  Analyse — Grouping and Aggregation




```python
airline_summary = df.groupby("Airline").agg({
    "Flight_ID": "count",
    "Passengers": "sum",
    "Average_Ticket_Price": "mean",
    "Passenger_Satisfaction": "mean"
})

airline_summary = airline_summary.rename(columns={
    "Flight_ID": "Number_of_Flights",
    "Passengers": "Total_Passengers",
    "Average_Ticket_Price": "Average_Ticket_Price",
    "Passenger_Satisfaction": "Average_Satisfaction"
})

airline_summary.round(2)
```

### Average delay by airline




```python
airline_delay = delay_df.groupby("Airline")["Delay_Minutes"].mean()

airline_delay = airline_delay.rename("Average_Delay")

airline_delay.sort_values(ascending=False).round(2)
```

## Flight Status Analysis


```python
status_summary = df.groupby("Flight_Status")["Flight_ID"].count()
status_summary
```

```python
status_percentage = (
    df["Flight_Status"].value_counts(normalize=True) * 100
).round(2)

status_percentage
```

##  Weather and Delay Analysis


```python
weather_summary = delay_df.groupby("Weather").agg({
    "Flight_ID": "count",
    "Delay_Minutes": "mean"
}).rename(columns={
    "Flight_ID": "Flights_With_Recorded_Delay",
    "Delay_Minutes": "Average_Delay"
})

weather_summary.sort_values(
    "Average_Delay",
    ascending=False
).round(2)
```

## Travel Class Analysis


```python
class_summary = df.groupby("Travel_Class").agg({
    "Flight_ID": "count",
    "Average_Ticket_Price": "mean",
    "Passenger_Satisfaction": "mean"
}).rename(columns={
    "Flight_ID": "Number_of_Flights",
    "Average_Ticket_Price": "Average_Ticket_Price",
    "Passenger_Satisfaction": "Average_Satisfaction"
})

class_summary.round(2)
```

##  Monthly Analysis


```python
monthly_summary = df.groupby("Flight_Month").agg({
    "Flight_ID": "count",
    "Passengers": "sum"
}).rename(columns={
    "Flight_ID": "Number_of_Flights",
    "Passengers": "Total_Passengers"
})

monthly_summary
```

##  Route Analysis




```python
route_summary = df.groupby(
    ["Origin", "Destination"]
).agg({
    "Flight_ID": "count",
    "Passengers": "sum"
}).rename(columns={
    "Flight_ID": "Number_of_Flights",
    "Passengers": "Total_Passengers"
})

route_summary.sort_values(
    "Number_of_Flights",
    ascending=False
).head(10)
```

## Transformation with `melt()`




```python
airline_metrics = airline_summary.reset_index()[[
    "Airline",
    "Average_Ticket_Price",
    "Average_Satisfaction"
]]

airline_metrics_long = pd.melt(
    airline_metrics,
    id_vars=["Airline"],
    var_name="Metric",
    value_name="Value"
)

airline_metrics_long.head(12)
```

##  Combining Data with `concat()`



```python
first_half = df.iloc[:90]
second_half = df.iloc[90:]

combined_df = pd.concat(
    [first_half, second_half],
    ignore_index=True
)

print("Original rows:", len(df))
print("Rows after concat:", len(combined_df))

combined_df.head()
```

##  Combining Data with `merge()`




```python
passenger_summary = (
    df.groupby("Airline")["Passengers"]
    .sum()
    .reset_index()
    .rename(columns={"Passengers": "Total_Passengers"})
)

delay_summary = (
    delay_df.groupby("Airline")["Delay_Minutes"]
    .mean()
    .reset_index()
    .rename(columns={"Delay_Minutes": "Average_Delay"})
)

merged_airline = pd.merge(
    passenger_summary,
    delay_summary,
    on="Airline"
)

merged_airline.sort_values(
    "Total_Passengers",
    ascending=False
).round(2)
```

##  Data Type and Memory Optimization




```python
df.memory_usage(deep=True)
```

### Convert repeated text columns to `category`




```python
optimized_df = df.copy()

category_columns = [
    "Airline",
    "Origin",
    "Destination",
    "Aircraft",
    "Travel_Class",
    "Flight_Status",
    "Weather",
    "Booking_Channel",
    "Meal_Preference",
    "Delay_Category"
]

for col in category_columns:
    optimized_df[col] = optimized_df[col].astype("category")

optimized_df.memory_usage(deep=True)
```

### Numeric downcasting




```python
for col in ["Passengers", "Seat_Capacity", "Average_Ticket_Price", "Avg_Baggage_Kg", "Passenger_Satisfaction"]:
    optimized_df[col] = pd.to_numeric(
        optimized_df[col],
        downcast="integer"
    )

optimized_df.dtypes
```

# Key Observations

1. **Most flights are On Time:** 114 out of 180 flights are On Time, while 59 are Delayed and 7 are Cancelled.

2. **Vistara has the most flight records:** Vistara appears in 37 of the 180 records, the highest among the airlines.

3. **IndiGo has the highest average recorded delay:** Its average recorded delay is about **25.03 minutes**, while Air India Express has the lowest at about **11.29 minutes**.

4. **Rain has the highest average recorded delay:** Among flights with a recorded delay, Rain has an average delay of about **22.22 minutes**.

5. **Premium Economy has the highest average passenger satisfaction:** Its average satisfaction is about **3.69/5**, higher than Economy and Business.

6. **Online Travel Portal is the most common booking channel:** It appears in **50 flights**, followed by Mobile App with 45.

7. **Delhi–Chennai is the most frequently recorded route:** It appears **17 times** in the dataset.

8. **The dataset contains data-quality issues:** 7 flights have missing delay values because they are cancelled, and 9 records have passengers greater than the recorded seat capacity.
