# Branch-wise Performance

```python

branch_performance = df.groupby("Branch")[[
    "Customers", "Revenue", "Profit", "Marketing_Spend",
    "Staff_Count", "Avg_Delivery_Min", "Customer_Rating"
]].mean().round(2)

branch_performance
```

```python

branch_performance.sort_values("Profit", ascending=False)
```

```python

branch_performance.sort_values("Revenue", ascending=False)
```

### Region-wise Performance

```python

region_performance = df.groupby("Region")[[
    "Customers", "Revenue", "Profit", "Marketing_Spend",
    "Avg_Delivery_Min", "Customer_Rating"
]].mean().round(2)

region_performance
```

```python

region_performance[["Revenue", "Profit"]].sort_values("Revenue", ascending=False)
```

## Store Type Analysis

```python

store_performance = df.groupby("Store_Type")[[
    "Customers", "Orders", "Average_Bill", "Revenue",
    "Profit", "Marketing_Spend", "Avg_Delivery_Min",
    "Customer_Rating"
]].mean().round(2)

store_performance
```

```python

store_performance.sort_values("Profit", ascending=False)
```

## Bivariate Analysis

```python
# Studying the relationship between important pairs of variables

print("Customers and Revenue:")
print(df["Customers"].corr(df["Revenue"]))

print("\nMarketing Spend and Profit:")
print(df["Marketing_Spend"].corr(df["Profit"]))

print("\nStaff Count and Customers:")
print(df["Staff_Count"].corr(df["Customers"]))

print("\nDelivery Time and Customer Rating:")
print(df["Avg_Delivery_Min"].corr(df["Customer_Rating"]))
```

## Correlation Matrix

```python
correlation_matrix = df[[
    "Staff_Count", "Marketing_Spend", "Customers", "Orders",
    "Average_Bill", "Revenue", "Food_Cost", "Operating_Cost",
    "Profit", "Avg_Delivery_Min", "Customer_Rating",
    "Temperature_C", "Humidity_Percent"
]].corr().round(2)

correlation_matrix
```

```python

profit_correlation = correlation_matrix["Profit"].sort_values(ascending=False)
profit_correlation
```

The correlation value lies between **-1 and +1**. Values closer to +1 show a positive relationship, values closer to -1 show a negative relationship, and values close to 0 show a weak linear relationship. Correlation shows association, not cause and effect.

## Final Observations

1. **Branch performance:** Bengaluru has the highest average profit of **₹31,890.91**, while Mumbai has the lowest average profit of **₹26,580.99**.

2. **Regional performance:** The **South** region has the highest average revenue at **₹88,320.22**.

3. **Store type:** **Premium** stores have the highest average profit at **₹62,890.24**, which is much higher than the other store types.

4. **Customer volume:** **Premium** stores have the highest average customers at **298.73** customers, while Express stores have the lowest average.

5. **Customers and revenue:** Customers and Revenue have a strong positive correlation of **0.88**. This means branches with more customers generally tend to have higher revenue.

6. **Marketing spend and profit:** Marketing Spend and Profit have a positive correlation of **0.35**. This shows a relationship between the two variables, but correlation does not prove that marketing spend directly causes higher profit.

7. **Staff and customers:** Staff Count and Customers have a positive correlation of **0.69**, suggesting that branches with more staff generally handle higher customer volumes.

8. **Delivery time and rating:** Average Delivery Time and Customer Rating have a negative correlation of **-0.44**. This suggests that longer delivery times are generally associated with lower ratings in this dataset.

## Conclusion

The EDA shows that restaurant performance varies across branches, regions and store types. Customer volume is closely related to revenue, while profit is also related to several operational factors such as customers, costs and marketing spend. The branch-wise and store-type comparisons help identify better-performing groups, while the correlation matrix gives a quick view of relationships between numerical variables.
