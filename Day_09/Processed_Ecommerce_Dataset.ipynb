# Processed E-commerce Dataset


```python
import pandas as pd

orders = pd.read_csv("Orders.csv")
customers = pd.read_csv("Customers.csv")
products = pd.read_csv("Products.csv")

print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
```

```python
orders.head()
```

```python
customers.head()
```

```python
products.head()
```

## Merge Orders and Customers

The `Customer_ID` column is common in both datasets.


```python
orders_customers = orders.merge(
    customers,
    on="Customer_ID",
    how="left"
)

orders_customers.head()
```

## Merge Product Information


```python
processed = orders_customers.merge(
    products,
    on="Product_ID",
    how="left"
)

processed.head()
```

## Using concat()




```python
orders_part1 = orders.iloc[:5]
orders_part2 = orders.iloc[5:10]

combined_orders = pd.concat([orders_part1, orders_part2])

combined_orders
```

## Convert Order Date



```python
processed["Order_Date"] = pd.to_datetime(processed["Order_Date"])

processed[["Order_ID", "Order_Date"]].head()
```

## Extract Month, Day and Day of Week

```python
processed["Month"] = processed["Order_Date"].dt.month
processed["Day"] = processed["Order_Date"].dt.day
processed["Day_of_Week"] = processed["Order_Date"].dt.day_name()

processed[[
    "Order_ID",
    "Order_Date",
    "Month",
    "Day",
    "Day_of_Week"
]].head()
```

## Using apply()

I used `apply()` to calculate the total amount for each order.

**Total Amount = Quantity × Unit Price**


```python
processed["Total_Amount"] = processed.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

processed[[
    "Order_ID",
    "Product_Name",
    "Quantity",
    "Unit_Price",
    "Total_Amount"
]].head(10)
```

## Create the Final Clean DataFrame

```python
processed = processed[
    [
        "Order_ID", "Order_Date", "Month", "Day", "Day_of_Week",
        "Customer_ID", "Customer_Name", "City", "Region",
        "Membership_Type", "Product_ID", "Product_Name",
        "Category", "Brand", "Unit_Price", "Quantity",
        "Total_Amount", "Payment_Method", "Order_Status"
    ]
]

processed.head()
```

## Export the Processed Dataset



```python
processed.to_csv(
    "Day9_Processed_Ecommerce_Dataset.csv",
    index=False
)

print("Processed dataset saved successfully.")
```

- The three datasets were successfully combined using `merge()`.
- Customer and product details are now available along with each order.

- The order date was converted to DateTime format and month, day and day of week were extracted.
- `apply()` was used to calculate the total amount for each order.
