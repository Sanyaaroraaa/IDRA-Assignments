# E-commerce Sales Analysis



```python
import pandas as pd

df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")
df.head()
```

```python
## Basic Information
print("Rows and Columns:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Selecting one column
df["Product"].head()
```

```python
# Selecting multiple columns
df[["Product", "Category", "Quantity", "Total_Sales"]].head()
```

### Filtering Data


```python
# Orders where sales are more than 5000
df[df["Total_Sales"] > 5000].head(10)
```

```python
# Orders from the Electronics category
df[df["Category"] == "Electronics"].head()
```

```python
# Orders where quantity is 3 or more
df[df["Quantity"] >= 3].head()
```

## Sorting Data

```python
# Highest sales first
df.sort_values("Total_Sales", ascending=False).head(10)
```

```python
# Lowest sales first
df.sort_values("Total_Sales", ascending=True).head(10)
```

##  Basic Sales Statistics

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print("Total Sales:", total_sales)
print("Average Sales:", round(average_sales, 2))
print("Highest Sale:", highest_sales)
print("Lowest Sale:", lowest_sales)
print("Total Quantity Sold:", total_quantity)
print("Number of Orders:", number_of_orders)
```

##  Category-wise Sales Analysis

```python
category_sales = df.groupby("Category")["Total_Sales"].sum()
category_sales.sort_values(ascending=False)
```

```python
# Category-wise quantity sold
category_quantity = df.groupby("Category")["Quantity"].sum()
category_quantity.sort_values(ascending=False)
```

```python
# Category-wise average sales
df.groupby("Category")["Total_Sales"].mean().sort_values(ascending=False)
```

## City-wise Sales Analysis

```python
city_sales = df.groupby("City")["Total_Sales"].sum()
city_sales.sort_values(ascending=False)
```

```python
# Cities with the highest number of orders
df.groupby("City")["Order_ID"].nunique().sort_values(ascending=False)
```

##  Product-wise Sales Analysis

```python
product_sales = df.groupby("Product")["Total_Sales"].sum()
product_sales.sort_values(ascending=False).head(10)
```

```python
# Top products by quantity sold
df.groupby("Product")["Quantity"].sum().sort_values(ascending=False).head(10)
```

##  Payment Method Analysis

```python
payment_sales = df.groupby("Payment_Method")["Total_Sales"].sum()
payment_sales.sort_values(ascending=False)
```

```python
# Number of orders for each payment method
df.groupby("Payment_Method")["Order_ID"].nunique().sort_values(ascending=False)
```

## Finding a Specific Order

```python
# Order with the highest sales
df[df["Total_Sales"] == df["Total_Sales"].max()]
```

```python
# Total sales, average sales and quantity for each category
df.groupby("Category").agg({
    "Total_Sales": ["sum", "mean", "max", "min"],
    "Quantity": "sum"
}).sort_values(("Total_Sales", "sum"), ascending=False)
```
